```python
import numpy as np
import pandas as pd
import seaborn as sns

import matplotlib.pyplot as plt
import os
import matplotlib as mpl
from matplotlib.colors import Normalize, LinearSegmentedColormap
from mpl_toolkits.axes_grid1 import make_axes_locatable
```

```python
def truncate_colormap(cmap, minval=0.0, maxval=1.0, n=256):
    new_cmap = LinearSegmentedColormap.from_list(
        f'trunc({cmap.name},{minval:.2f},{maxval:.2f})',
        cmap(np.linspace(minval, maxval, n)))
    return new_cmap

from matplotlib.colors import Normalize

def plot_scatter(data,prop,prop2,prop_color,min_scale,max_scale,ax,color_map,marker,marker_size,zorder,prop_error_var=None,z=None):
    norm = Normalize(vmin=min_scale, vmax=max_scale)
    color = color_map(norm(data[prop_color].values))

    if(prop_error_var is None):
        ax.scatter(data[prop].values,data[prop2].values,marker=marker,s=marker_size,zorder=zorder,c=color)
    else:
        ax.errorbar(data[prop].values,data[prop2].values,z * np.sqrt(data[prop_error_var].values),marker=marker, linestyle='',markersize=marker_size,zorder=zorder,c=color)
```

```python
# Define once, use everywhere
standard_linewidth = 1.5  # Choose a value that looks good

spinesParams = {
    'axes.spines.right': True,
    'axes.spines.top': True,
    'axes.linewidth': standard_linewidth,
}



tex_fonts = {
    # Use LaTeX to write all text
    "text.usetex": True,
    "font.family": "serif",
    # Use 10pt font in plots, to match 10pt font in document
    "axes.labelsize": 10,
    "font.size": 10,
    # Make the legend/label fonts a little smaller
    "legend.fontsize": 8,
    "xtick.labelsize": 8,
    "ytick.labelsize": 8
}
tickParams = {
    "xtick.top":True,
    "xtick.bottom":True,
    "xtick.direction": "in",
    "ytick.left":True,
    "ytick.right":True,
    "ytick.direction": "in",
}
spinesParams = {
    'axes.spines.right' : True,
    'axes.spines.top' : True
}
plt.rcParams.update(tex_fonts)
plt.rcParams.update(tickParams)
plt.rcParams['text.latex.preamble'] = r'\usepackage{amsmath}\usepackage{amssymb}'
plt.rcParams.update(spinesParams)  # ADD THIS LINE

inchPerCm = 0.393701
goldenRatio = (1+np.sqrt(5))/2
figWidthCm = 17.9
figHeightCm = figWidthCm/1.2
#figHeightCm = figWidthCm*1.2

figWidthInches = figWidthCm*inchPerCm
figHeightInches = figHeightCm*inchPerCm

colors=sns.color_palette("colorblind")
markers=['o','s','<','v','h','P','*','^','8','p','H','d','D','X']
marker_size=5

lw=1
lw2=1
```

```python
fig_label_x=-0.15
fig_label_y=1
```

```python
color_map_genta = truncate_colormap(mpl.colormaps['Blues'], 0.3, 1.0)
color_map_cipro = mpl.colormaps['Wistia']  # Keep as is
color_map_chp = truncate_colormap(mpl.colormaps['Greens'], 0.3, 1.0)
color_map_tetra = truncate_colormap(mpl.colormaps['OrRd'], 0.3, 1.0)


max_scale_genta=xmax[0]
max_scale_cipro=xmax[1]
max_scale_chp=xmax[3]
max_scale_tetra=xmax[2]

color_map_lambda=mpl.colormaps['Blues']
color_map_r=mpl.colormaps['OrRd']


min_scale_lambda=0.5
max_scale_lambda=10


min_scale_r=1
max_scale_r=6
```

```python
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
from mpl_toolkits.axes_grid1 import make_axes_locatable

fig = plt.figure(figsize=(figWidthInches, figHeightInches))
gs = fig.add_gridspec(2, 2, height_ratios=[1, 1], hspace=0.4, wspace=0.1)

#  TOP ROW 
ax0 = fig.add_subplot(gs[0, :])
marker_size = 50

ax0.plot([0,1],[0,1],'gray', lw=1)
for i,d in enumerate(Genta.dataset.unique()):
    plot_scatter(Genta[Genta.dataset==d], 'prob_neg_drop','q_chip_like','concentration',
                  0, max_scale_genta, ax0, color_map_genta, 'o', marker_size, i)
for i,d in enumerate(Cipro.dataset.unique()):
    plot_scatter(Cipro[Cipro.dataset==d], 'prob_neg_drop','q_chip_like','concentration',
                  0, max_scale_cipro, ax0, color_map_cipro, 's', marker_size, i)
for i,d in enumerate(Chp.dataset.unique()):
    plot_scatter(Chp[Chp.dataset==d], 'prob_neg_drop','q_chip_like','concentration',
                  0, max_scale_chp, ax0, color_map_chp, 'v', marker_size, i)
for i,d in enumerate(Tetra.dataset.unique()):
    plot_scatter(Tetra[Tetra.dataset==d], 'prob_neg_drop','q_chip_like','concentration',
                  0, max_scale_tetra, ax0, color_map_tetra, '*', marker_size, i)

# Legend text
ax0.text(0.642, 0.4, r'$\bullet$ Gentamicin $\left[\frac{\mu g}{ml}\right]$', transform=ax0.transAxes, color=colors[0])
ax0.text(0.64, 0.3, r'$\blacksquare$ Ciprofloxacin $\left[\frac{n g}{ml}\right]$', transform=ax0.transAxes, color=colors[1])
ax0.text(0.642, 0.2, r'$\blacktriangledown$ Chloramphenicol $\left[\frac{\mu g}{ml}\right]$', transform=ax0.transAxes, color=colors[2])
ax0.text(0.638, 0.1, r'$\bigstar$ Tetracycline $\left[\frac{\mu g}{ml}\right]$', transform=ax0.transAxes, color=colors[3])

ax0.set_xlabel(r'$d_{neg}$')
ax0.set_ylabel('q', rotation=0)
ax0.set_xlim([0,1])
ax0.set_ylim([0,1])

#  BOTTOM ROW 
ax1 = fig.add_subplot(gs[1, 0])
ax2 = fig.add_subplot(gs[1, 1])
ax = [ax1, ax2]
marker_size = 25

# lambda_like (ax[0])
for i,d in enumerate(Genta.dataset.unique()):
    plot_scatter(Genta[Genta.dataset==d], 'prob_neg_drop','q_chip_like','lambda_like',
                  min_scale_lambda, max_scale_lambda, ax[0], color_map_lambda,'o', marker_size,i)
for i,d in enumerate(Cipro.dataset.unique()):
    plot_scatter(Cipro[Cipro.dataset==d], 'prob_neg_drop','q_chip_like','lambda_like',
                  min_scale_lambda, max_scale_lambda, ax[0], color_map_lambda,'s', marker_size,i)
for i,d in enumerate(Chp.dataset.unique()):
    plot_scatter(Chp[Chp.dataset==d], 'prob_neg_drop','q_chip_like','lambda_like',
                  min_scale_lambda, max_scale_lambda, ax[0], color_map_lambda,'v', marker_size,i)
for i,d in enumerate(Tetra.dataset.unique()):
    plot_scatter(Tetra[Tetra.dataset==d], 'prob_neg_drop','q_chip_like','lambda_like',
                  min_scale_lambda, max_scale_lambda, ax[0], color_map_lambda,'*', marker_size,i)

ax[0].plot([0,1],[0,1],'gray', lw=1)

# r_like (ax[1])
for i,d in enumerate(Genta.dataset.unique()):
    plot_scatter(Genta[Genta.dataset==d], 'prob_neg_drop','q_chip_like','r_like',
                  min_scale_r, max_scale_r, ax[1], color_map_r,'o', marker_size,i)
for i,d in enumerate(Cipro.dataset.unique()):
    plot_scatter(Cipro[Cipro.dataset==d], 'prob_neg_drop','q_chip_like','r_like',
                  min_scale_r, max_scale_r, ax[1], color_map_r,'s', marker_size,i)
for i,d in enumerate(Chp.dataset.unique()):
    plot_scatter(Chp[Chp.dataset==d], 'prob_neg_drop','q_chip_like','r_like',
                  min_scale_r, max_scale_r, ax[1], color_map_r,'v', marker_size,i)
for i,d in enumerate(Tetra.dataset.unique()):
    plot_scatter(Tetra[Tetra.dataset==d], 'prob_neg_drop','q_chip_like','r_like',
                  min_scale_r, max_scale_r, ax[1], color_map_r,'*', marker_size,i)

ax[0].set_xlabel(r'$d_{neg}$')
ax[0].set_ylabel('q', rotation=0,va='center')
ax[1].set_xlabel(r'$d_{neg}$')
ax[1].set_ylabel('q', rotation=0,va='center')
for a in ax:
    a.set_xlim([0,1])
    a.set_ylim([0,1])

#  layout adjustment 
plt.tight_layout(w_pad=0.1,h_pad=0.1,rect=[0,0,0.75,1])

# Get axes positions after tight_layout
pos0 = ax0.get_position()
pos1 = ax[0].get_position()
pos2 = ax[1].get_position()

#  colorbars 
cbar_width = 0.015
spacing = 0.05
start_x = pos0.x1 + 0.02
colorbar_end_x = start_x + 3*(cbar_width + spacing) + cbar_width

# Top row colorbars
cax_g = fig.add_axes([start_x, pos0.y0, cbar_width, pos0.height])
cb_g = plt.colorbar(plt.cm.ScalarMappable(norm=Normalize(0,max_scale_genta), cmap=color_map_genta), cax=cax_g)
cax_c = fig.add_axes([start_x + (cbar_width+spacing), pos0.y0, cbar_width, pos0.height])
cb_c = plt.colorbar(plt.cm.ScalarMappable(norm=Normalize(0,max_scale_cipro), cmap=color_map_cipro), cax=cax_c)
cax_ch = fig.add_axes([start_x + 2*(cbar_width+spacing), pos0.y0, cbar_width, pos0.height])
cb_ch = plt.colorbar(plt.cm.ScalarMappable(norm=Normalize(0,max_scale_chp), cmap=color_map_chp), cax=cax_ch)
cax_t = fig.add_axes([start_x + 3*(cbar_width+spacing), pos0.y0, cbar_width, pos0.height])
cb_t = plt.colorbar(plt.cm.ScalarMappable(norm=Normalize(0,max_scale_tetra), cmap=color_map_tetra), cax=cax_t)

#  bottom row positions
vertical_gap = 0.08
bottom_gap = 0.1
bottom_cbar_gap = 0.005
label_space = 0.025  # Space for the 'r' label to match top row end

# Calculate total width including label space
total_width = colorbar_end_x - pos0.x0 + label_space
plot_width = (total_width - bottom_gap - 2*cbar_width - 2*bottom_cbar_gap) / 2

# Reposition bottom plots with vertical gap
ax[0].set_position([pos0.x0, pos0.y0 - pos0.height - vertical_gap, plot_width, pos1.height])
ax[1].set_position([pos0.x0 + plot_width + bottom_gap, pos0.y0 - pos0.height - vertical_gap, plot_width, pos2.height])

# Update positions after resizing
pos1 = ax[0].get_position()
pos2 = ax[1].get_position()

# Bottom colorbars (right of each plot)
cax1 = fig.add_axes([pos1.x1 + bottom_cbar_gap, pos1.y0, cbar_width, pos1.height])
cb1 = plt.colorbar(plt.cm.ScalarMappable(norm=Normalize(min_scale_lambda, max_scale_lambda), cmap=color_map_lambda), cax=cax1)
cb1.set_label(r'$\tilde{\Lambda}$', rotation=0, ha='left', labelpad=-3,va='center')

cax2 = fig.add_axes([pos2.x1 + bottom_cbar_gap, pos2.y0, cbar_width, pos2.height])
cb2 = plt.colorbar(plt.cm.ScalarMappable(norm=Normalize(min_scale_r, max_scale_r), cmap=color_map_r), cax=cax2)
cb2.set_label('r', rotation=0, ha='left', labelpad=-3,va='center')

label_x = -0.11
ax0.text(-0.06, 1.04, '(a)', transform=ax0.transAxes, fontsize=12, va='bottom')
ax1.text(label_x, 1.04, '(b)', transform=ax1.transAxes, fontsize=12, va='bottom')
ax2.text(label_x, 1.04, '(c)', transform=ax2.transAxes, fontsize=12, va='bottom')

plt.show()
```
