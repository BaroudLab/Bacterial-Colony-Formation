```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

import os
from AnalyseInitialCondition import * 
```

```python
def histogram_ini(data,fit_parameters,xlim,ylim,ax,line_width):

    
    initial_cells=np.arange(0,np.max(data)+1,dtype=int)
    bin_count=np.bincount(data)
    total_samples=np.sum(bin_count)
        
    ax.bar(initial_cells,bin_count/np.sum(bin_count),color='#7ac2f4')
    ax.errorbar(initial_cells,bin_count/total_samples,np.sqrt(bin_count/total_samples**2) * (1-bin_count/total_samples),elinewidth=2,fmt='None',ecolor='#359fe8',zorder=4)
        
    ax.plot(initial_cells, poisson.pmf(initial_cells,fit_parameters.lam),color='#8b63ed',lw=line_width , label='Poisson')
    ax.plot(initial_cells, nbinom.pmf(initial_cells,fit_parameters.r_like,fit_parameters.r_like/(fit_parameters.r_like+fit_parameters.lam)), color='#ad053f',lw=line_width , label='Negative binomial')
 

    ax.set_xlim([-0.4,xlim])
    ax.set_ylim([0,ylim])
```

```python
tex_fonts = {
    # Use LaTeX to write all text
    "text.usetex": True,
    "font.family": "serif",
    # Use 10pt font in plots, to match 10pt font in document
    "axes.labelsize": 10,
    "font.size": 10,
    # Make the legend/label fonts a little smaller
    "legend.fontsize": 8,
    "xtick.labelsize": 8,
    "ytick.labelsize": 8
}
tickParams = {
    "xtick.top":True,
    "xtick.bottom":True,
    "xtick.direction": "in",
    "ytick.left":True,
    "ytick.right":True,
    "ytick.direction": "in",
}
spinesParams = {
    'axes.spines.right' : True,
    'axes.spines.top' : True
}
plt.rcParams.update(tex_fonts)
plt.rcParams.update(tickParams)
inchPerCm = 0.393701


lw=1
marker_size=10
lw2=1
```

# Gentamicin

```python
Genta = pd.read_csv(f'Tables/Gentamicin_measured.csv',index_col=0)
Genta_fit=pd.read_csv(f'Tables/Gentamicin_ini_fit.csv',index_col=0)
```

```python
figWidthCm = 17.9
figHeightCm =figWidthCm

figWidthInches = figWidthCm*inchPerCm
figHeightInches = figHeightCm*inchPerCm





dataset=0
conc=Genta[Genta.dataset==dataset].concentration.unique()

#Genta dataset 0

match dataset:
    case 0:
        xlim=[10,10,10,10,12,10]
        ylim=[0.4,0.4,0.4,0.4,0.4,0.45]

    case 1:
        xlim=[10,10,10,10,10,10]
        ylim=[0.8,0.8,0.8,0.8,0.8,0.8]  
    case 2:
        xlim=[20,20,20,20,20,45]
        ylim=[0.4,0.4,0.4,0.4,0.4,0.15]




fig, ax = plt.subplots(3,2, figsize=(figWidthInches, figHeightInches))

for row,a_rows in enumerate(ax):
    for col,a in enumerate(a_rows):
        i=row * a_rows.shape[0] + col
        histogram_ini(Genta[(Genta.dataset==dataset) & (Genta.concentration==conc[i])].n_cells.values,Genta_fit[(Genta_fit.dataset==dataset) & (Genta_fit.concentration==conc[i])],xlim[i],ylim[i],a,lw)
        
        label=conc[i]
        if(label.is_integer()):
            label=int(label)
        ab_str=f'Gentamicin {label}'
        a.text(0.3,0.9, ab_str + r'$\left[\frac{\mu g}{ml}\right]$', transform=a.transAxes)


ax[0,0].legend(frameon=False,fancybox=True,handlelength=1.5,bbox_to_anchor=(1,0.8))
        
ax[0,0].set_ylabel('')
ax[0,1].set_xlabel('')

ax[1,0].set_ylabel('probability')
ax[1,1].set_xlabel('')

ax[2,0].set_ylabel('')
ax[2,1].set_xlabel('')


plt.tight_layout(w_pad=0.6)


ax[2,0].text(0.9,-0.2, 'initial detected cells', transform=ax[2,0].transAxes)


plt.subplots_adjust(left=0.07,bottom=0.07)

plt.show()
```
